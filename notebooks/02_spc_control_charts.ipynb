```python
import sys, os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, "C:/Users/sharanu.ganji/secom-yield-dashboard/src")
from spc import compute_i_mr, compute_cusum, western_electric_rules, plot_i_mr, plot_cusum

DATA_DIR = "C:/Users/sharanu.ganji/secom-yield-dashboard/data"
os.makedirs(os.path.join(DATA_DIR, "spc_plots"), exist_ok=True)

# Load cleaned data
df = pd.read_csv(os.path.join(DATA_DIR, "secom_cleaned.csv"), parse_dates=["Time"])
df = df.sort_values("Time").reset_index(drop=True)
print(f"Loaded: {df.shape}")

# Pick top 5 sensors by variance
sensor_cols = [c for c in df.columns if c.startswith("X")]
top_sensors = df[sensor_cols].var().nlargest(5).index.tolist()
print(f"Top 5 sensors: {top_sensors}")

# I-MR on top sensor
sensor = top_sensors[0]
values = df[sensor].values
times = df["Time"].values
stats = compute_i_mr(values)
print(f"\n{sensor}: mean={stats['mean']:.4f}, sigma={stats['sigma']:.4f}")
plot_i_mr(times, values, stats, title=f"I-MR: {sensor}",
          save_path=os.path.join(DATA_DIR, "spc_plots", f"i_mr_{sensor}.png"))

# CUSUM
cusum_stats = compute_cusum(values)
print(f"\nCUSUM signals: {len(cusum_stats['signals'])}")
plot_cusum(times, values, cusum_stats, title=f"CUSUM: {sensor}",
           save_path=os.path.join(DATA_DIR, "spc_plots", f"cusum_{sensor}.png"))

# Western Electric
violations = western_electric_rules(values)
print(f"\nWestern Electric violations: {len(violations)}")
if violations:
    vdf = pd.DataFrame(violations, columns=["index", "rule", "description"])
    print(vdf.head(10))

# 2nd sensor
sensor2 = top_sensors[1]
values2 = df[sensor2].values
stats2 = compute_i_mr(values2)
plot_i_mr(times, values2, stats2, title=f"I-MR: {sensor2}",
          save_path=os.path.join(DATA_DIR, "spc_plots", f"i_mr_{sensor2}.png"))

print("\n✅ SPC analysis complete.")   
```
